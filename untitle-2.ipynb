#Numpy is faster than python list and array see with example below

```python
import numpy as np 
import time

#python list example 

n = 1000000
list1 = list(range(n)) #python list 
start_time = time.time()

lst_1 = [ x * 2  for x in list1]

end_time = time.time()
print("Python list time:",end_time - start_time) 

#numpy example 
arr = np.arange(n)

start = time.time()
arr = arr * 2 
end = time.time()

print("Numpy time:",end - start)
```

# indexing and slicing 

```python
import numpy as np 

arr = np.array([1,2,3,4,5])
print(arr[4]) #acessing last element 
print(arr[0]) #accesing first element 

```

```python
#negative indexing  

print(arr[-2])
```

```python
#slicing  

print(arr[1:4]) #accessing element form 1 to 3 
```

```python
#reshaping 
#when you want convert one dimensional array 
#into two dimensional array 

import numpy as np 
arr = np.arange(1,13) 
print(arr)
```

```python
new_arr = arr.reshape(3,4) #here 3 rows and 4 comumns 
print(new_arr)
```
