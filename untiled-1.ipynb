```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

student = {
    "name": "Akshay Bane",
    "email": "akshaybane4@gmail.com",
    "username": "akshay24"
}

response = requests.post(url,json=student)
print(response.json())
print(response.status_code)
```

```python
#when you updating user always needs id after endpoints 

import requests

url = "https://jsonplaceholder.typicode.com/users/1"

student_update = {
    "name": "Akshay Bane",
    "email": "akshaybane4@example.com",
    "username": "akshay55"
}

response = requests.put(url, json=student_update)
print(response.json())
print(response.status_code)
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/users/1"

response = requests.delete(url)

print(response.status_code)
```
